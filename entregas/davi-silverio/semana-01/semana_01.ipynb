```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dados = pd.read_csv('cleaned.csv', header=None)
dados.head()
```

```python
#Estatísticas de alt: média, máximo, mínimo e número de amostras
alt = dados[3]

media = alt.mean()
maximo = alt.max()
minimo = alt.min()
amostras = alt.count()

print(f"Média: {media:.2f}")
print(f"Máximo: {maximo:.2f}")
print(f"Mínimo: {minimo:.2f}")
print(f"Número de amostras: {amostras}")
```

```python
#plote ax, ay e az contra o tempo (millis) no mesmo gráfico
dados["tempo_s"] = (dados[1] - dados[1].min()) / 1000
dados.plot(x="tempo_s", y=[10, 11, 12], title="Aceleração (ax, ay, az) X Tempo", figsize=(12, 6), grid=True)
```

```python
#plote vz (velocidade vertical) e alt (altitude) ao longo do tempo
dados["velocidade vertical"] = dados[13]
dados["altitude"] = dados[3]

dados.plot(x="tempo_s", y="velocidade vertical", figsize=(12, 5), color='r', grid=True, title='Gráfico de velocidade vertical')

dados.plot(x="tempo_s", y="altitude", figsize=(12, 5), color='g', grid=True, title='Gráfico de altitude')
```
